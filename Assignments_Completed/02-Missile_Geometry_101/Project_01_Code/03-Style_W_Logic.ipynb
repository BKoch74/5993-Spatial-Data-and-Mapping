## 03 - Geojson Properties

### Goal:
Manipulate feature properties dynamically.

### Task:
- Load the world GeoJSON. 
- Color **only countries whose name starts with A–M** one color.
- Color **N–Z** another color.
- This means you will be checking `feature["properties"]` for each country. 
- No manual hardcoding of country names.

```python
from pathlib import Path

cwd = Path.cwd()

target = cwd / "data" / "countries.geojson"
print(target)

exists = target.exists()
print("\nDoes file exist?")
print(exists)

assert exists, (
    f"\n❌ ERROR: File not found:\n{target}\n"
    "Check spelling and folder structure."
)
```

```python
import json
from rich import print
import folium



with open(target,"r") as f:
   data = json.load(f)
   
print(data["features"][0])
    
gj_type = data.get("type", None)
features = data.get("features", [])
print("GeoJSON type:", gj_type)
print("Feature count:", len(features))

assert gj_type == "FeatureCollection", f"❌ Expected FeatureCollection, got: {gj_type}"
assert len(features) > 0, "❌ FeatureCollection has zero features."


# Create an instance of a map to put data on
m = folium.Map(location=[0,0], zoom_start=3)
  
# Display map without changing colors
folium.GeoJson(
    data,
    name="world"
).add_to(m)
  
m


  
```

```python
def style_function(feature):
    name = feature["properties"].get("ADMIN", "")
    first = name[0].lower() if name else "n"
    if first in "abcdefghijklm":
        return {
            "fillColor": "#3498DB",
            "color": "#1A5276",
            "weight": 0.8,
            "fillOpacity": 0.65,
        }
    else:
        return {
            "fillColor": "#E74C3C",
            "color": "#922B21",
            "weight": 0.8,
            "fillOpacity": 0.65,
        }

m2 = folium.Map(location=[0, 0], zoom_start=3)

folium.GeoJson(
    data,
    name="world",
    style_function=style_function
).add_to(m2)

m2
```
